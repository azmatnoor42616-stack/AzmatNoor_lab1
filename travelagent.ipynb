```python
print("TRAVEL AGENT")

places = {
    "skardu": 80000,
    "hunza": 60000,
    "fairy meadows": 50000
}

city = input("Where do you want to travel?: ").lower().strip()
budget = int(input("How much is your budget?: "))

if city in places:
    price = places[city]

    if budget >= price:
        print("Agent: You can afford a trip to", city.title())
    else:
        print("Agent: Sorry, this trip is too expensive.")
else:
    print("Agent: Sorry, we don't have trips to that place.")
```
