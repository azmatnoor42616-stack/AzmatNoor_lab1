```python
!pip install -U google-genai
from google import genai

client = genai.Client(api_key="")


MODEL = "models/gemini-3.5-flash"

fuuast_dictionary = {"who is teaching AI to BSCS 6 semester section B":"Dr. Uzma Afzal is currently teaching AI to BSCS 6 semester section B", "who is teaching CC to BSCS 6 semester section B":"Miss Naheed Azeem is currently teaching CC to BSCS 6 semester section B","who is teaching NLP to BSCS 6 semester section B":"Dr. Sarim is currently teaching CC to BSCS 6 semester section B", "who is teaching NSC to BSCS 6 semester section B":"Dr. Syed Akhtar Raza is currently teaching NSC to BSCS 6 semester section B", "what is the monday class schedule for BSCS 6 semester section B":"AI > 12:20, NLP > 1:00, NSC > 2:00", "what is the tuesday class schedule for BSCS 6 semester section B":"AI > 11:30, NSC > 12:20", "what is the wednesday class schedule for BSCS 6 semester section B":"AI > 12:20, NSC > 1:10", "what is the thursday class schedule for BSCS 6 semester section B":"NSC > 9:00, PP > 11:30, CC > 12:20, NLP > 1:10", "what is monday class schedule of BSCS 6 semester section B":"AI > 12:20 then NLP > 1:00 then NSC > 2:00", "what is the monday class schedule for BSCS 6 semester section B":"AI > 12:20 then NLP > 1:00 then NSC > 2:00"}
def fuuast_agent(user_input):

    prompt = f"""
You are a agent who have all the information about fedral Urdu University of Arts, Science and technology.whatever user asks you, you will answered in a very good way that user understands it clearly and if the user asks you any specific question then you will check it in dictionary which i provided you.
If user asks any specific question use this dictionary to answer:{fuuast_dictionary}
Make response very clear that user understand everything.

User request: {user_input}

About FUUAST:
"""

    response = client.models.generate_content(
        model=MODEL,
        contents=prompt
    )

    return response.text



print("Gemini FUUAST Agent Started (type 'exit' to stop)\n")

while True:
    user = input("You: ")

    if user.lower() == "exit":
        print("Goodbye ")
        break

    try:
        reply = fuuast_agent(user)
        print("\nAgent:\n", reply, "\n")

    except Exception as e:
        print("Error:", e)
```
